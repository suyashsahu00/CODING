```python
import matplotlib.pyplot as plt

# 1. Define Data
x = ["Python", "C", "C++", "Java"]
y = [70, 50, 40, 60]   # Popularity scores

# 2. Create Horizontal Bar Chart
plt.barh(x, y, color="y", label="Popularity")

# 3. Add Labels and Title
plt.xlabel("Number (Score)", fontsize=15)
plt.ylabel("Language", fontsize=15)
plt.title("Prog Lang Comparison", fontsize=15)

# 4. Add Legend and Show
plt.legend()
plt.show()
```
